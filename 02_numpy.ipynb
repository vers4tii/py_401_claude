# Partie 2 — NumPy

**Objectif** : calculer vite sur beaucoup de nombres, sans écrire de boucle.

NumPy est la bibliothèque de calcul numérique de Python. Son objet central est le **tableau** (`array`) : une liste de nombres, du même type, sur laquelle on peut faire des calculs en une seule ligne. Pandas, que l'on verra ensuite, est construit au-dessus de NumPy.

````python
import numpy as np 
````

## 1. Pourquoi NumPy ? Liste contre tableau

Avec une liste, `* 2` **répète** la liste. Avec un tableau NumPy, `* 2` **multiplie chaque valeur**.

````python
import numpy as np

liste = [1, 2, 3]
tableau = np.array([1, 2, 3])

print(liste * 2)     # [1, 2, 3, 1, 2, 3]
print(tableau * 2)   # [2 4 6]

# Avec une liste, il faudrait une boucle :
print([x * 2 for x in liste])
````

Et c'est beaucoup plus rapide. Comparons sur un million de nombres :

````python
import time

grande_liste = list(range(1_000_000))
grand_tableau = np.arange(1_000_000)

t = time.perf_counter()
resultat = [x * 2 for x in grande_liste]
print(f"Liste   : {time.perf_counter() - t:.4f} s")

t = time.perf_counter()
resultat = grand_tableau * 2
print(f"NumPy   : {time.perf_counter() - t:.4f} s")
````

## 2. Créer des tableaux

````python
print(np.array([10, 20, 30]))      # à partir d'une liste
print(np.arange(0, 10, 2))         # de 0 à 10 (exclu), de 2 en 2
print(np.linspace(0, 1, 5))        # 5 valeurs régulières entre 0 et 1
print(np.zeros(4))                 # quatre zéros
print(np.ones(3))                  # trois uns
````

## 3. Calculer sur tout le tableau

Les opérations s'appliquent **élément par élément**, sans boucle : on dit que le calcul est **vectorisé**.

````python
prix = np.array([2.0, 3.5, 4.0, 1.3])
quantites = np.array([10, 4, 6, 20])

print(prix * 1.2)            # +20 % sur chaque prix
print(prix * quantites)      # montant de chaque ligne
print((prix * quantites).sum())   # total
````

## 4. Les statistiques de base

````python
temperatures = np.array([12, 15, 14, 18, 21, 19, 16])

print("Moyenne :", temperatures.mean())
print("Médiane :", np.median(temperatures))
print("Écart-type :", temperatures.std().round(2))
print("Min / Max :", temperatures.min(), temperatures.max())
print("Position du max :", temperatures.argmax())   # indice du jour le plus chaud
````

## 5. Sélectionner des valeurs

L'indexation marche comme pour les listes. Mais NumPy a une arme en plus : le **masque**. On écrit une condition, NumPy renvoie un tableau de `True` / `False`, et on s'en sert pour filtrer.

````python
temperatures = np.array([12, 15, 14, 18, 21, 19, 16])

print(temperatures[0], temperatures[-1], temperatures[2:5])

masque = temperatures > 16
print(masque)                    # [False False False  True  True  True False]
print(temperatures[masque])      # seulement les valeurs > 16
print(masque.sum())              # combien de True (True compte pour 1)

# np.where : choisir une valeur selon une condition
print(np.where(temperatures > 16, "chaud", "frais"))
````

## 6. Les tableaux à 2 dimensions

Un tableau peut avoir des **lignes et des colonnes**, comme un tableur. `shape` donne ses dimensions.

Le paramètre `axis` dit dans quel sens on calcule :
- `axis=0` : on calcule **en descendant**, colonne par colonne ;
- `axis=1` : on calcule **en traversant**, ligne par ligne.

````python
# 3 magasins (lignes) × 4 semaines (colonnes)
ventes = np.array([
    [120, 135, 150, 160],   # Paris
    [ 80,  90,  85, 100],   # Lyon
    [ 60,  70,  75,  65],   # Marseille
])
print(ventes.shape)              # (3, 4) : 3 lignes, 4 colonnes
print(ventes.sum(axis=1))        # total par magasin (une valeur par ligne)
print(ventes.sum(axis=0))        # total par semaine (une valeur par colonne)
print(ventes[0])                 # la ligne de Paris
print(ventes[:, 2])              # la 3e semaine pour tous les magasins
````

### Exercice 5 · Niveau 3 · défi · en binôme

**Défi** : simulez **1 000 lancers** d'un dé à 6 faces et affichez combien de fois chaque face est sortie.
Indices : `rng = np.random.default_rng(42)` puis `rng.integers(1, 7, size=1000)` ; ensuite `np.unique(..., return_counts=True)`.
Bonus : affichez la fréquence en pourcentage. Est-elle proche de 16,7 % ?

````python
rng = np.random.default_rng(42)
lancers = rng.integers(1, 7, size=1000)

faces, counts = np.unique(lancers, return_counts=True)
for face, count in zip(faces, counts):
    print(f"Face {face} : {count} occurrences")

# Bonus : fréquence en %
freqs = counts / lancers.size * 100
for face, f in zip(faces, freqs):
    print(f"Face {face} : {f:.1f} %")
````
