## 1. Les variables

Une variable est une **boîte avec un nom** dans laquelle on range une valeur. Le signe `=` veut dire « range la valeur de droite dans la boîte de gauche ».

Les 4 types de base :

| Type | Exemple | Sert à |
| --- | --- | --- |
| `int` | `12` | nombre entier |
| `float` | `3.5` | nombre à virgule (on écrit un **point**) |
| `str` | `"Paris"` | texte, entre guillemets |
| `bool` | `True` ou `False` | vrai ou faux |

```python
prenom = "Lina"
age = 21
taille = 1.68
est_etudiante = True

print(prenom, age, taille, est_etudiante)
print(type(age), type(taille), type(prenom), type(est_etudiante))
```

Pour mélanger du texte et des variables, on utilise une **f-string** : un `f` devant les guillemets, et les variables entre accolades `{ }`.

```python
print(f"{prenom} a {age} ans.")
print(f"Dans 10 ans, {prenom} aura {age + 10} ans.")
prix = 19.987
print(f"Prix arrondi : {prix:.2f} €")   # :.2f = 2 chiffres après la virgule
```

## 2. Les listes

Une liste range **plusieurs valeurs dans l'ordre**, entre crochets `[ ]`.
Attention : en Python, on compte à partir de **0**. Le premier élément est `liste[0]`, le dernier est `liste[-1]`.

```python
villes = ["Paris", "Lyon", "Marseille"]

print(villes[0])        # le premier
print(villes[-1])       # le dernier
print(len(villes))      # le nombre d'éléments

villes.append("Lille")  # ajouter à la fin
print(villes)

print(villes[1:3])      # du 2e au 3e (l'indice 3 est exclu)
```

```python
notes = [12, 15, 8, 17, 10]
print("Somme :", sum(notes))
print("Plus haute :", max(notes))
print("Plus basse :", min(notes))
print("Triées :", sorted(notes))
```

## 3. Les dictionnaires

Un dictionnaire range des **paires clé → valeur**, entre accolades `{ }`. C'est comme un vrai dictionnaire : on cherche un mot (la clé) pour trouver sa définition (la valeur).

```python
stock = {
    "croissant": 40, 
    "cookie": 12, 
    "muffin": 3
    }

print(stock["cookie"])       # lire une valeur
stock["muffin"] = 10         # modifier une valeur
stock["cheesecake"] = 6      # ajouter une paire
print(stock)

print("croissant" in stock)  # la clé existe-t-elle ?
print(list(stock.keys()))
print(list(stock.values()))
```

## 4. Les conditions

`if` exécute un bloc **seulement si** la condition est vraie. `elif` veut dire « sinon, si… », `else` veut dire « sinon ».

**L'indentation compte** : les lignes décalées de 4 espaces appartiennent au bloc.

Comparaisons : `==` égal, `!=` différent, `<`, `<=`, `>`, `>=`. On combine avec `and`, `or`, `not`.

```python
note = 13

if note >= 16:
    print("Très bien")
elif note >= 10:
    print("Admis")
else:
    print("Refusé")
```

## 5. Les boucles

`for` répète un bloc **pour chaque élément** d'une liste. `range(n)` produit les nombres de 0 à n-1.

```python
for ville in ["Paris", "Lyon", "Marseille"]:
    print("Magasin de", ville)

for i in range(3):
    print("Tour numéro", i)

# Parcourir un dictionnaire
stock = {"croissant": 40, "cookie": 12, "muffin": 3}
for produit, quantite in stock.items():
    print(produit, "→", quantite)
```

### Exercice 4 · Niveau 2 · moyen · en binôme

Avec `notes = [12, 15, 8, 17, 10, 9]`, affichez pour chaque note :
`12 → admis` ou `8 → refusé` (admis si la note est supérieure ou égale à 10).
Puis affichez le **nombre d'admis**.

```python
# Votre code ici
notes = [12, 15, 8, 17, 10, 9]
admis = 0
for note in notes:
    if note >= 10:
        print(f"{note} → admis")
        admis += 1
    else:
        print(f"{note} → refusé")

print(f"Nombre d'admis : {admis}")
```

## 6. Les fonctions

Une fonction est un **morceau de code réutilisable** qui porte un nom. On la définit avec `def`, elle reçoit des **paramètres** et renvoie un résultat avec `return`.

On écrit une fonction dès qu'on se répète.

```python
def moyenne(valeurs):
    return sum(valeurs) / len(valeurs)

print(moyenne([12, 15, 8]))
print(moyenne([20, 18]))

# Un paramètre peut avoir une valeur par défaut
def saluer(prenom, langue="fr"):
    if langue == "en":
        return f"Hello {prenom}!"
    return f"Bonjour {prenom} !"

print(saluer("Lina"))
print(saluer("Lina", langue="en"))
```
